```python
import pandas as pd

df = pd.read_csv("../data/orders_dataset.csv")

# 1. Total row count
total_rows = len(df)
print(f"Total rows: {total_rows}")

# 2. Overall return rate
return_rate = df['returned'].mean()
print(f"Overall return rate: {return_rate:.4f} ({return_rate*100:.2f}%)")

# 3. Missing rating_given
missing_pct = df['rating_given'].isna().mean()
print(f"Missing rating_given: {missing_pct:.4f} ({missing_pct*100:.2f}%)")

# 4. Return rate by product_category
print("\nReturn rate by product_category:")
print(df.groupby('product_category')['returned'].mean().sort_values(ascending=False))

# 5. Return rate by payment_method
print("\nReturn rate by payment_method:")
print(df.groupby('payment_method')['returned'].mean().sort_values(ascending=False))

# 6. Missingness of rating_given by product_category and payment_method
categorical_cols = ['product_category', 'payment_method']

for col in categorical_cols:
    print(f"\nMissing rating_given rate by {col}:")
    print(df.groupby(col)['rating_given'].apply(lambda x: x.isna().mean()).sort_values(ascending=False))

# 7. Missingness of rating_given by payment_method (the MAR evidence)
print("\nMissing rating_given rate by payment_method:")
print(df.groupby('payment_method')['rating_given'].apply(lambda x: x.isna().mean()).sort_values(ascending=False))
```

About 13 % of 'rating given' values were missing.While comparing the missing rate across product category and payment method, the missingness in rating_given is not random but explained by another observed column, making it Missing at random, explained by payment method, where COD customers pay on delivery and often skip the rating step after delivery of product
