```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split

```

```python
# Load the dataset
df =pd.read_csv('../data/orders_dataset.csv')
x = df.drop('returned', axis=1)
y = df['returned'] #target column
print(df.shape)
print(df.dtypes)
print (f'Number of null values in each column:\n{df.isnull().sum()}')
```

```python
# Train-test split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size =0.2, random_state = 42)
print(f"Training set shape: {x_train.shape}")
print(f"Testing set shape: {x_test.shape}")
```

```python
#imputing missing values and encoding categorical variables
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()),])
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))])
numeric_features = ['price_inr', 'rating_given', 'delivery_days', 'discount_pct', 'customer_tenure_days', 
                    'num_previous_orders','num_previous_returns', 'delivery_distance_km', 'is_weekend_order']
categorical_features = ['product_category','payment_method']
preprocessor= ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ],
    remainder='drop'
)
print("number of numeric features:", len(numeric_features))
print("number of categorical features:", len(categorical_features))
print("total number of features:", len(numeric_features) + len(categorical_features))
print("total number of features in dataset:", x_train.shape[1])
```

```python
#training
preprocessor.fit(x_train)
x_train_transformed = preprocessor.transform(x_train)
x_test_transformed = preprocessor.transform(x_test)

```

```python
#sanity check
print("Shape of transformed training data:", x_train_transformed.shape)
print("Shape of transformed testing data:", x_test_transformed.shape)
print("feature names after transformation:", preprocessor.get_feature_names_out())
```
